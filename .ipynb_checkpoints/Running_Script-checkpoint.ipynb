```python
case=1

t = np.linspace(0, 100, 10000)/ cst.tfs

FWHM  = 20. / cst.tfs
sigma = FWHM/(2.0*np.sqrt(2*np.log(2)))
E0 = 1.7

efield = Hamiltonian.E(t, sigma,E0, ω=0.5/cst.Ry, case=case )
#plotting.plot_electric_pulse(t*cst.tfs, efield, np.array([[0,100],[-0.002*0,0.002]]),fout='')

J     = 0.5/cst.Ry
delta = 0.5/cst.Ry
k_list=[0, 0.25, 0.50, 0.75]


t_span= (t[0],t[-1])


solutions=np.zeros((len(k_list),2,len(t)),dtype=np.complex64)
for k_i,k in enumerate(k_list):
    y0 = np.array([1.0, 0.0, 0.0, 0.0], dtype=complex)
    sol = solve_ivp(lambda t, y: Hamiltonian.rhs(t, y, k, J, delta, sigma, E0, ω=0.5/cst.Ry,case=case), t_span, y0, t_eval=t)
    solutions[k_i,0,:] = sol.y[0]
    solutions[k_i,1,:] = sol.y[1]
```

```python
limits=np.array([[0,100],[0.0,0.2]])
plotting.plot_population_field(t*cst.tfs,efield,k_list,solutions.real,limits)
```

### After fixing a case, test the DMD algorithm

```python
case=1

t = np.linspace(0, 100, 10000)/ cst.tfs

FWHM  = 20. / cst.tfs
sigma = FWHM/(2.0*np.sqrt(2*np.log(2)))
E0 = 1.7

efield = Hamiltonian.E(t, sigma,E0, ω=0.5/cst.Ry, case=case )
#plotting.plot_electric_pulse(t*cst.tfs, efield, np.array([[0,100],[-0.002*0,0.002]]),fout='')

J      = 0.5/cst.Ry
delta  = 0.5/cst.Ry
k_list = np.linspace(-np.pi, np.pi, 101)[:-1]


t_span= (t[0],t[-1])


solutions=np.zeros((len(k_list),2,len(t)),dtype=np.complex64)
for k_i,k in enumerate(k_list):
    y0 = np.array([1.0, 0.0, 0.0, 0.0], dtype=complex)
    sol = solve_ivp(lambda t, y: Hamiltonian.rhs(t, y, k, J, delta, sigma, E0, ω=0.5/cst.Ry,case=case), t_span, y0, t_eval=t)
    solutions[k_i,0,:] = sol.y[0]
    solutions[k_i,1,:] = sol.y[1]
```

```python
import numpy as np
import matplotlib.pyplot as plt
from pydmd.preprocessing import hankel_preprocessing

# 1. Generate synthetic data
t = np.linspace(0, 10, 200)
X = np.sin(2 * np.pi * 0.5 * t) + 0.3 * np.sin(2 * np.pi * 1.5 * t)
X = X.reshape(1, -1)  # Shape: (1, 200)

# 2. Custom Hankel embedding function
def custom_hankel_embedding(X, delay):
    """
    Create Hankel matrix from data matrix X (shape n x m)
    with 'delay' number of rows in the embedding.
    Returns: (delay*n) x (m - delay + 1)
    """
    n, m = X.shape
    hankel_matrix = np.vstack([X[:, i: m - delay + i + 1] for i in range(delay)])
    return hankel_matrix

delay = 10

# Apply custom Hankel embedding
H_custom = custom_hankel_embedding(X, delay)

# 3. Apply PyDMD's preprocessing
from pydmd import DMD
dmd = DMD(svd_rank=0)
delay_dmd = hankel_preprocessing(dmd, d=delay)
delay_dmd.fit(X)

# PyDMD's internal Hankel matrix
H_pydmd = delay_dmd.snapshots

# 4. Compare both matrices
print("Are they approximately equal?", np.allclose(H_custom, H_pydmd))

# Optional: visualize the result
plt.figure(figsize=(10, 4))
plt.subplot(1, 3, 1)
plt.imshow(H_custom, aspect='auto')
plt.title("Custom Hankel Embedding")
plt.colorbar()

plt.subplot(1, 3, 2)
plt.imshow(H_pydmd, aspect='auto')
plt.title("PyDMD Hankel Embedding")
plt.colorbar()

plt.subplot(1, 3, 3)
plt.imshow(H_custom-H_pydmd, aspect='auto')
plt.title("Diff")
plt.colorbar()

plt.tight_layout()
plt.show()
```

```python
d = 10  # we will use this number of delays throughout the tutorial
rank=4

#---------------------------------------------------------------

dmd = DMD(svd_rank=rank)
delay_dmd = hankel_preprocessing(dmd, d=d)
delay_dmd.fit(X.T)

#---------------------------------------------------------------
import scipy.linalg as LA

def custom_hankel_embedding(X, delay):
    n, m = X.shape
    hankel_matrix = np.vstack([
        X[:, i: m - delay + i + 1] for i in range(delay)
    ])
    return hankel_matrix

def perform_dmd(H, rank=None, d=1, dt=1):
    #print(np.shape(H))
    X1 = H[:, :-1]
    X2 = H[:, 1:]

    U, S, Vh = svd(X1, full_matrices=False)

    if rank is not None:
        U_r  = U[:, :rank]
        S_r  = S[:rank]
        Vh_r = Vh[:rank, :]
    else:
        U_r  = U[:, :]
        S_r  = S[:]
        Vh_r = Vh[:, :]

    Sigma_inv = np.diag(1 / (S_r + 0.000001 ))
    A_tilde = U_r.T @ X2 @ Vh_r.T @ Sigma_inv
    eigvals, W = LA.eig(A_tilde)
    Phi = X2 @ Vh_r.T @ Sigma_inv @ W
    if d > 1:
        Φ = np.average( Phi.reshape(d,Phi.shape[0] // d, Phi.shape[1],),axis=0,)
    
    

    

    return S,eigvals,Phi, Φ
    #x0 = X1[:, 0]
    #b = pinv(Phi) @ x0
    #omega = np.log(eigvals) / dt
    #timesteps = np.arange(X1.shape[1]) * dt + t0
    #print(timesteps)
    #time_dynamics = np.array([b * np.exp(omega * t) for t in timesteps]).T

    #X_dmd = Phi @ time_dynamics
    #return eigvals, Phi, X_dmd.real, timesteps


def Dynamics(H,Eigvals,Φ,nt,d):
    x1=H[:,0]

    time_steps = np.linspace(0,nt-d,nt-d+1,dtype=int)
    temp = np.repeat( Eigvals[:, None], time_steps.shape[0], axis=1 )
    tpow = ( time_steps - 0 ) / ( time_steps[1] - time_steps[0] )

    amplitudes = LA.pinv(Φ) @ x1
    dynamics = np.power(temp, tpow) * amplitudes[:, None]

    return temp, amplitudes, dynamics , np.dot(Φ,dynamics)


H_custom = custom_hankel_embedding(X.T, d)
Singular_values,Eigvals,Modes,modes_average = perform_dmd(H_custom, rank=rank, d=d )
temp_costum,amplitudes_costum,b_costum,reconstruction_costum = Dynamics(H_custom,Eigvals,Modes,nt,d)

print(np.shape(H_custom))


#---------------------------------------------------------------

print(np.shape(delay_dmd.modes))
print(np.argsort(-np.abs(dmd.amplitudes)),dmd.amplitudes,dmd.eigs)


mode_order = np.argsort(-np.abs(dmd.amplitudes))
lead_eigs = dmd.eigs[mode_order]
lead_modes = dmd.modes[:, mode_order]
lead_dynamics = dmd.dynamics[mode_order]
print("dmd_timesteps:",dmd.dmd_timesteps.shape[0])
print("original_time:",dmd.original_time)
print(dmd._opt)

print(np.shape(lead_modes))
#lead_modes = dmd.modes

if d > 1:
    lead_modes = np.average( lead_modes.reshape(d,lead_modes.shape[0] // d, lead_modes.shape[1],),axis=0,)
    #lead_modes = np.average( lead_modes.reshape(lead_modes.shape[0] // d,d, lead_modes.shape[1],),axis=1,)
print(np.shape(lead_modes))


temp = np.repeat(
            dmd.eigs[:, None], dmd.dmd_timesteps.shape[0], axis=1
        )



#-----------------------------------------------------

plt.figure()
plt.plot(x,lead_modes[:,0].real)
plt.plot(x,modes_average[:,0].real)

plt.figure()
plt.plot(lead_dynamics[3].real)
plt.plot(b_costum[3].real)

plt.figure()
plt.plot(amplitudes_costum.real,amplitudes_costum.imag,'o')
plt.plot(dmd.amplitudes.real,dmd.amplitudes.imag,'o')
```
