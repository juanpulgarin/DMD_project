### This section shows what happens when changing the different parameters in your Hamiltonian

```python
k_vals = np.linspace(-np.pi, np.pi, 101)[:-1]

J_list     = np.array([1.0])/cst.Ry
delta_list = np.array([0.5,1.5])/cst.Ry

valence_bands    = {}
conduction_bands = {}
for J in J_list:
    for Δ in delta_list:
        valence_bands[J,Δ]    = Hamiltonian.eps_vb( k_vals, J, Δ )
        conduction_bands[J,Δ] = Hamiltonian.eps_cb( k_vals, J, Δ )

limits=np.array([[-np.pi,np.pi],[-3.5,3.5]])

plotting.plot_band_structure(k_vals, J_list, delta_list, valence_bands, conduction_bands, limits,fout='')

for J in J_list:
    for Δ in delta_list:
        plotting.plot_band_structure(k_vals, np.array([J]), np.array([Δ]), valence_bands, conduction_bands, limits,fout='')



J_list     = np.array([1.0,2.5])/cst.Ry
delta_list = np.array([0.5])/cst.Ry


valence_bands    = {}
conduction_bands = {}
for J in J_list:
    for Δ in delta_list:
        valence_bands[J,Δ]    = Hamiltonian.eps_vb( k_vals, J, Δ )
        conduction_bands[J,Δ] = Hamiltonian.eps_cb( k_vals, J, Δ )

limits=np.array([[-np.pi,np.pi],[-3.5,3.5]])

plotting.plot_band_structure(k_vals, J_list, delta_list, valence_bands, conduction_bands, limits,fout='')

for J in J_list:
    for Δ in delta_list:
        plotting.plot_band_structure(k_vals, np.array([J]), np.array([Δ]), valence_bands, conduction_bands, limits,fout='')
```

```python
k_vals = np.linspace(-np.pi, np.pi, 101)[:-1]

J_list     = np.array([0.5])/cst.Ry
delta_list = np.array([0.5])/cst.Ry

valence_bands    = {}
conduction_bands = {}
for J in J_list:
    for Δ in delta_list:
        valence_bands[J,Δ]    = Hamiltonian.eps_vb( k_vals, J, Δ )
        conduction_bands[J,Δ] = Hamiltonian.eps_cb( k_vals, J, Δ )

limits=np.array([[-np.pi,np.pi],[-2.5,2.5]])

plotting.plot_band_structure(k_vals, J_list, delta_list, valence_bands, conduction_bands, limits,fout='')
```

```python
case=1

t = np.linspace(0, 100, 10000)/ cst.tfs

FWHM  = 20. / cst.tfs
sigma = FWHM/(2.0*np.sqrt(2*np.log(2)))
E0 = 2.7

efield = Hamiltonian.E(t, sigma,E0, ω=0.5/cst.Ry, case=case )
#plotting.plot_electric_pulse(t*cst.tfs, efield, np.array([[0,100],[-0.002*0,0.002]]),fout='')

J     = 0.5/cst.Ry
delta = 0.5/cst.Ry
#k_list=[0,0.05,0.1,0.15,0.2,0.25,0.3,0.35,0.4,0.45,0.5,0.55,0.6,0.65,0.7,0.75,0.8]
k_list=np.linspace(0,0.75,40)


t_span= (t[0],t[-1])


solutions=np.zeros((len(k_list),2,len(t)),dtype=np.complex64)
for k_i,k in enumerate(k_list):
    y0 = np.array([1.0, 0.0, 0.0, 0.0], dtype=complex)
    sol = solve_ivp(lambda t, y: Hamiltonian.rhs(t, y, k, J, delta, sigma, E0, ω=0.5/cst.Ry,case=case), t_span, y0, t_eval=t)
    solutions[k_i,0,:] = sol.y[0]
    solutions[k_i,1,:] = sol.y[1]
```

```python
limits=np.array([[0,100],[0.0,0.2]])
plotting.plot_population_field(t*cst.tfs,efield,k_list,solutions.real,limits)
```

### After fixing a case, test the DMD algorithm

```python
case=1

t = np.linspace(0, 100, 10000)/ cst.tfs

FWHM  = 20. / cst.tfs
sigma = FWHM/(2.0*np.sqrt(2*np.log(2)))
E0 = 1.7

efield = Hamiltonian.E(t, sigma,E0, ω=0.5/cst.Ry, case=case )
#plotting.plot_electric_pulse(t*cst.tfs, efield, np.array([[0,100],[-0.002*0,0.002]]),fout='')

J      = 0.5/cst.Ry
delta  = 0.5/cst.Ry
k_list = np.linspace(-np.pi, np.pi, 101)[:-1]


t_span= (t[0],t[-1])


solutions=np.zeros((len(k_list),2,len(t)),dtype=np.complex64)
for k_i,k in enumerate(k_list):
    y0 = np.array([1.0, 0.0, 0.0, 0.0], dtype=complex)
    sol = solve_ivp(lambda t, y: Hamiltonian.rhs(t, y, k, J, delta, sigma, E0, ω=0.5/cst.Ry,case=case), t_span, y0, t_eval=t)
    solutions[k_i,0,:] = sol.y[0]
    solutions[k_i,1,:] = sol.y[1]
```

```python
limits=np.array([[0,100],[0.0,0.2]])
plotting.plot_population_field(t*cst.tfs,efield,k_list,solutions.real,limits)
```

```python
X = solutions[:,0,:-1]
Y = solutions[:,0,1: ]

truncamiento=-6

#eigvals, Phi_L, r_L, S      = Hamiltonian.perform_dmd(X, Y, t, 3)
mu, Phi_M, r_M, Sig2 = Hamiltonian.DMD(X, Y,orden_truncado=[0,truncamiento])
```

```python
plotting.plot_singular_values(Sig2,10**(truncamiento))
plotting.plot_eigenvalues(mu)
```

```python
X = solutions[:,0,:-1]
Y = solutions[:,0,1: ]

truncamiento=-6


mu, Phi_M, r_M, Sig2 = Hamiltonian.DMD(X, Y, orden_truncado=[0,truncamiento])

# Time evolution from DMD output
dt = t[1] - t[0]
omega_M = np.log(mu) / dt
b_M = np.linalg.pinv(Phi_M) @ X[:, 0]
time_dynamics_M = np.array([b_M * np.exp(omega_M * T) for T in t]).T
X_dmd_M = Phi_M @ time_dynamics_M
```

```python
from pydmd.plotter import plot_eigs, plot_summary

def f1(x, t):
    return 1.0 / np.cosh(x + 3) * np.cos(2.3 * t)


def f2(x, t):
    return 2.0 / np.cosh(x) * np.tanh(x) * np.sin(2.8 * t)


nx = 65  # number of grid points along space dimension
nt = 129  # number of grid points along time dimension

# Define the space and time grid for data collection.
x = np.linspace(-5, 5, nx)
t = np.linspace(0, 4 * np.pi, nt)
xgrid, tgrid = np.meshgrid(x, t)
dt = t[1] - t[0]  # time step between each snapshot

# Data consists of 2 spatiotemporal signals.
X1 = f1(xgrid, tgrid)
X2 = f2(xgrid, tgrid)
X = X1 + X2

# Make a version of the data with noise.
mean = 0
std_dev = 0.2
random_matrix = np.random.normal(mean, std_dev, size=(nt, nx))
Xn = X + random_matrix
```

```python
d = 10  # we will use this number of delays throughout the tutorial
rank=4

#---------------------------------------------------------------

dmd = DMD(svd_rank=rank)
delay_dmd = hankel_preprocessing(dmd, d=d)
delay_dmd.fit(X.T)

mode_order = np.argsort(-np.abs(dmd.amplitudes))
lead_eigs = dmd.eigs[mode_order]
lead_modes = dmd.modes[:, mode_order]
lead_dynamics = dmd.dynamics[mode_order]

#---------------------------------------------------------------

H_custom = dmd_scripts.custom_hankel_embedding(X.T, d)
Singular_values,Eigvals,Modes,modes_average = dmd_scripts.perform_dmd(H_custom, rank=rank, d=d )
temp_costum,amplitudes_costum,b_costum,reconstruction_costum = dmd_scripts.Dynamics(H_custom,Eigvals,Modes,nt,d)

```

```python
plt.figure(figsize=(12, 12))

plt.subplot(3, 2, 1)
plt.title("Eigenvalues: Custom vs PyDMD")
plt.plot(lead_eigs.real, lead_eigs.imag, 'go', label='PyDMD')
plt.plot(Eigvals.real, Eigvals.imag, 'r*', label='Custom')

plt.xlabel("Real")
plt.ylabel("Imaginary")
plt.grid(True)
plt.legend()

plt.subplot(3, 2, 2)
plt.title("Singular Values")
plt.plot(np.log(Singular_values[:]/Singular_values[0]), 'g.', label='Custom')

plt.xlabel("Real")
plt.ylabel("Imaginary")
plt.grid(True)
plt.legend()

plt.subplot(3, 2, 3)
plt.title("Eigenmodes")
plt.plot(dmd.amplitudes.real,dmd.amplitudes.imag, 'go', label='PyDMD')
plt.plot(amplitudes_costum.real, amplitudes_costum.imag, 'r*', label='Custom')
plt.xlabel("Real")
plt.ylabel("Imaginary")
plt.grid(True)
plt.legend()

plt.subplot(3, 2, 4)
plt.imshow(np.dot(modes_average,b_costum).real)

plt.subplot(3, 2, 5)
plt.imshow(delay_dmd.reconstructed_data[0:65].real)

plt.subplot(3, 2, 6)
plt.imshow(reconstruction_costum[:,0:120].real)
```

```python
plotting.plot_singular_values(Singular_values,rank)
plotting.plot_eigenvalues(Eigvals)
plotting.plot_amplitudes_b(amplitudes_costum)
plotting.plot_modes_vs_grid(modes_average,0,np.array([[-5,5],[-0.1,0.1]]),x,ky=None,dimension=1)
plotting.plot_modes_vs_time(b_costum,0,np.array([[0,120],[-8,8]]))
plotting.plot_comparison_data_1d(X.T,reconstruction_costum,20,np.array([[0,129],[-1.3,1.3]]))

plotting.plot_comparison_data_2d(X.T,reconstruction_costum,np.array([[0,129],[-1.3,1.3]]))
```
