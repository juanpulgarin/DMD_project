```python
import numpy as np
import h5py
import matplotlib.pyplot as plt

import mosqDMD_class as dmd_lib
import Plotting as plotting
```

```python
fpath = "./../DATA/"
fname =  fpath + "WSe2-ExperimentalWindows.h5"

f = h5py.File(fname, "r")
k_exp  = np.array(f['Kexp_window'])
kp_exp = np.array(f['Kpexp_window'])
time = np.array(f['Kz_AtomicUnits'])


f.close()
```

```python
k_exp_t  = np.reshape(k_exp,[151*251,22] )
kp_exp_t = np.reshape(kp_exp,[151*251,22])


k_exp_total = np.vstack((k_exp_t, kp_exp_t))
```

```python
dmd_rank = 10
delay_embedding_dim = 10 # Set to 0 for no Hankel embedding
amplitude_calculation_method = 'least_outer' # Changed to 'least_outer' for demonstration
#frequencies_for_omega_reconstruction = np.linspace(-2*np.pi, 2*np.pi, 1000)
num_future_steps_to_reconstruct = 0

dmd = dmd_lib.DMD(rank=dmd_rank, delay_embedding_dim=delay_embedding_dim,tikhonov_regularization=None,\
                  amplitudes_method=amplitude_calculation_method)
dmd.fit(k_exp_t, time) # Pass the time array to the fit method
```

```python
plotting.plot_singular_values(dmd.S,dmd_rank)
plotting.plot_eigenvalues(dmd.eigvals)
```

```python
kx = np.linspace(0,151,152)[:-1]
ky = np.linspace(0,251,252)[:-1]
```

```python
plotting.plot_modes_vs_grid(np.reshape(dmd.Phi_physical,[151,251,dmd_rank]),0,np.array([[0,151],[0,251]]),kx,ky,dimension=2)
plotting.plot_modes_vs_grid(np.reshape(dmd.Phi_physical,[151,251,dmd_rank]),1,np.array([[0,151],[0,251]]),kx,ky,dimension=2)
plotting.plot_modes_vs_grid(np.reshape(dmd.Phi_physical,[151,251,dmd_rank]),2,np.array([[0,151],[0,251]]),kx,ky,dimension=2)
plotting.plot_modes_vs_grid(np.reshape(dmd.Phi_physical,[151,251,dmd_rank]),3,np.array([[0,151],[0,251]]),kx,ky,dimension=2)
plotting.plot_modes_vs_grid(np.reshape(dmd.Phi_physical,[151,251,dmd_rank]),4,np.array([[0,151],[0,251]]),kx,ky,dimension=2)
#plotting.plot_modes_vs_grid(np.reshape(dmd.Phi_physical,[151,251,dmd_rank]),5,np.array([[0,151],[0,251]]),kx,ky,dimension=2)
```
